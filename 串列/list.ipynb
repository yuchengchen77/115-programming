# 簡介
list是一個可以儲存多個元素(整數、福點數、字串)的容器
每個元素用逗號隔開並用中括號{}

```python
scores=[78,29,56,80,99]
print(scores)
```

## 索引值 
我們可以使用索引值(index)來存取串列中的元素
注意:索引值從0開始。



```python
print(scores[0]) #印出第1個元素
print(scores[1]) #印出第2個元素
```

## 串列內容的個數
len(a)取得a這個串列中元素的個數
len是length(長度)的意思

```python
print(len(scores))
print(scores[4]) #取得最後一個元素
print(scores[5-1])
print(scores[len(scores)-1])
print(scores[-1])
```
